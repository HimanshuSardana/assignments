**Getting quotes and author names both**

```python
import requests
from bs4 import BeautifulSoup

url = "https://quotes.toscrape.com"
response = requests.get(url)
soup = BeautifulSoup(response.text, "html.parser")

quotes = [q.text for q in soup.find_all("span", class_="text")]
authors = [a.text for a in soup.find_all("small", class_="author")]

for q, a in zip(quotes, authors):
    print(f"{q} — {a}")
```

**Handle Multiple Pages**

Most websites have pagination. We can loop through pages

```python
page = 1
while True:
    url = f"http://quotes.toscrape.com/page/{page}/"
    response = requests.get(url)
    soup = BeautifulSoup(response.text, "html.parser")

    quotes = soup.find_all("span", class_="text")
    if not quotes:
        break  # stop when no more quotes

    for quote in quotes:
        print(quote.get_text())

    page += 1
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

all_quotes = []

page = 1
while True:
    url = f"http://quotes.toscrape.com/page/{page}/"
    response = requests.get(url)
    soup = BeautifulSoup(response.text, "html.parser")

    quotes = soup.find_all("span", class_="text")
    authors = soup.find_all("small", class_="author")

    if not quotes:
        break

    for quote, author in zip(quotes, authors):
        all_quotes.append({
            "quote": quote.get_text(),
            "author": author.get_text()
        })

    page += 1

# Save to CSV and Excel
df = pd.DataFrame(all_quotes)
df.to_csv("quotes.csv", index=False)
df.to_excel("quotes.xlsx", index=False)
```

**Scraping JavaScript-heavy Websites**

Some websites don’t show full data in the raw HTML (they load content dynamically with JavaScript). In such cases, requests + BeautifulSoup won’t work directly.
We need a tool that can simulate a browser.

```python
pip install selenium
```

**Download a browser driver** (e.g., ChromeDriver) and make sure it matches your browser version.

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

url = "https://realpython.github.io/fake-jobs/"
response = requests.get(url)
soup = BeautifulSoup(response.text, "html.parser")

jobs = []

for card in soup.find_all("div", class_="card-content"):
    title = card.find("h2", class_="title").text.strip()
    company = card.find("h3", class_="company").text.strip()
    location = card.find("p", class_="location").text.strip()
    date_posted = card.find("time").text.strip()

    jobs.append({
        "title": title,
        "company": company,
        "location": location,
        "date_posted": date_posted
    })

# Save to CSV
df = pd.DataFrame(jobs)
df.to_csv("jobs.csv", index=False)

print(df.head())
```

**Scraping JavaScript-heavy websites is trickier than static HTML ones**, because the content you want often doesn’t exist in the raw HTML — it’s generated dynamically by JavaScript after the page loads.

There are 3 main approaches:

 **1. Use Selenium (Browser Automation)**

**Best for: Full JS-heavy sites (infinite scroll, logins, dynamic content).**

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
import time

# Setup Chrome options for headless mode
options = webdriver.ChromeOptions()
options.add_argument('--headless')
options.add_argument('--no-sandbox')
options.add_argument('--disable-dev-shm-usage')
options.add_argument('--disable-gpu') # Add this line
options.add_argument('--window-size=1920,1080') # Add this line

# Start browser using webdriver_manager
service = Service(ChromeDriverManager().install())
driver = webdriver.Chrome(service=service, options=options)

driver.get("http://quotes.toscrape.com/js/")  # JS-only quotes page
time.sleep(3)  # wait for JS to load

quotes = driver.find_elements(By.CLASS_NAME, "text")
authors = driver.find_elements(By.CLASS_NAME, "author")

for q, a in zip(quotes, authors):
    print(f"{q.text} — {a.text}")

driver.quit()
```

✅ Pro: Can handle logins, clicks, scrolls, JS rendering.
❌ Con: Slower (runs a whole browser).

**2. Use Playwright (Faster Browser Automation)**

Best for: Production scrapers that need speed & headless browsing.

```python
!pip install playwright
!playwright install
```

```python
#we need to use Playwright's asynchronous API.
import asyncio
from playwright.async_api import async_playwright

async def scrape_quotes():
    async with async_playwright() as p:
        browser = await p.chromium.launch(headless=True)
        page = await browser.new_page()
        await page.goto("http://quotes.toscrape.com/js/")
        await page.wait_for_timeout(3000)  # wait for JS
        #telling Playwright to pause the execution for 3000 milliseconds (or 3 seconds).
        #This is often used when dealing with JavaScript-heavy websites to give the page enough time to load and render content dynamically before attempting to scrape it.

        quotes = await page.query_selector_all(".text")
        authors = await page.query_selector_all(".author")
        # Since query_selector_all is an asynchronous operation (it involves waiting for the browser),
        #await is used to pause the execution until the operation is complete and the list of elements is returned.

        for q, a in zip(quotes, authors):
            print(f"{await q.inner_text()} — {await a.inner_text()}")

        await browser.close()

await scrape_quotes()
```

**Pro:** Faster than Selenium, modern API.
**Con:** Slightly more setup.

3. **Look for API Calls in Network Tab**

Sometimes, sites fetch data with an XHR / JSON API request. Instead of scraping rendered HTML, you can directly hit that API.

Example: Open DevTools → Network → XHR/Fetch → copy request URL.

**To find an actual API URL for the data you are interested in, you would typically:**

Go to the website you want to scrape in your web browser.
Open your browser's developer tools (usually by pressing F12).
Go to the "Network" tab.
Reload the page.
Filter the requests by "XHR" or "Fetch" to see the API calls being made.
Look for requests that seem to be retrieving the data you are interested in.
Copy the URL of that request.

**Webscraping the Books website**

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

base_url = "https://books.toscrape.com/"
url = base_url + "catalogue/page-1.html"

books = []

while True:
    response = requests.get(url)
    soup = BeautifulSoup(response.text, "html.parser")

    # find all books on the page
    items = soup.find_all("article", class_="product_pod")

    for item in items:
        title = item.h3.a["title"]
        price = item.find("p", class_="price_color").text.strip()
        availability = item.find("p", class_="instock availability").text.strip()
        rating = item.find("p", class_="star-rating")["class"][1]  # e.g. "Three"

        books.append({
            "title": title,
            "price": price,
            "availability": availability,
            "rating": rating
        })

    # find next page
    next_btn = soup.find("li", class_="next")
    if next_btn:
        next_url = next_btn.a["href"]
        url = base_url + "catalogue/" + next_url
    else:
        break

# save results
df = pd.DataFrame(books)
df.to_csv("books.csv", index=False)

print("Scraped", len(df), "books")
print(df.head())
```
